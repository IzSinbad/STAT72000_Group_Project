## Data Loading and Libraries

First, let's import the necessary libraries and load our dataset.

```python
# Load the dataset with the correct encoding
file_path = r"C:\Users\Adilf\Downloads\Most Streamed Spotify Songs 2024.csv (1)\Most Streamed Spotify Songs 2024.csv"
df = pd.read_csv(file_path, encoding='latin1')

# Display the first few rows of the dataset
df.head()
```

## Data Exploration and Cleaning

Let's explore the dataset and perform necessary cleaning operations.

```python
# Check basic information about the dataset
print("Dataset shape:", df.shape)
print("\nDataset info:")
df.info()
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Function to convert string-formatted numbers with commas to numeric
def convert_to_numeric(column):
    if column.dtype == 'object':
        return pd.to_numeric(column.str.replace(',', ''), errors='coerce')
    return column

# List of columns that should be numeric
numeric_columns = [
    'Spotify Streams', 'Spotify Playlist Count', 'Spotify Playlist Reach',
    'YouTube Views', 'YouTube Likes', 'TikTok Posts', 'TikTok Likes', 'TikTok Views',
    'YouTube Playlist Reach', 'AirPlay Spins', 'SiriusXM Spins',
    'Deezer Playlist Reach', 'Pandora Streams', 'Pandora Track Stations',
    'Soundcloud Streams', 'Shazam Counts', 'All Time Rank'
]

# Apply the conversion function to numeric columns
for col in numeric_columns:
    df[col] = convert_to_numeric(df[col])

# Check the data types after conversion
print("\nData types after conversion:")
df.select_dtypes(include=['float64', 'int64']).columns.tolist()
```

```python
# Convert release date to datetime and extract month and year
df['Release Date'] = pd.to_datetime(df['Release Date'], errors='coerce')
df['Release Year'] = df['Release Date'].dt.year
df['Release Month'] = df['Release Date'].dt.month
df['Release Day'] = df['Release Date'].dt.day

# Check the extracted date components
print("\nDate components:")
print(df[['Release Date', 'Release Year', 'Release Month', 'Release Day']].head())
```

```python
# Create platform ratios and other derived metrics
# YouTube to Spotify ratio
df['YouTube_Spotify_Ratio'] = df['YouTube Views'] / df['Spotify Streams']

# TikTok to Spotify ratio
df['TikTok_Spotify_Ratio'] = df['TikTok Views'] / df['Spotify Streams']

# Playlist reach per playlist (Spotify)
df['Spotify_Reach_Per_Playlist'] = df['Spotify Playlist Reach'] / df['Spotify Playlist Count']

# Engagement ratio for YouTube (likes per view)
df['YouTube_Engagement'] = df['YouTube Likes'] / df['YouTube Views']

# Engagement ratio for TikTok (likes per view)
df['TikTok_Engagement'] = df['TikTok Likes'] / df['TikTok Views']

# Check the derived metrics
print("\nDerived metrics:")
print(df[['YouTube_Spotify_Ratio', 'TikTok_Spotify_Ratio', 'Spotify_Reach_Per_Playlist', 
          'YouTube_Engagement', 'TikTok_Engagement']].describe())
```

```python
# Save the cleaned dataframe to a pickle file for use in other notebooks
df.to_pickle('cleaned_music_data.pkl')
print("Cleaned data saved to 'cleaned_music_data.pkl'")
```
